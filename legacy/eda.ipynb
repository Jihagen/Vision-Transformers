```python
# Jupyter cell

import os, json, pickle
from collections import Counter, defaultdict
from pathlib import Path

RESULTS_BASE = Path("../results")
PERSONA_DIR_PATTERN = "gdv_25_{i}"   # directories: gdv_25_0 ... gdv_25_9
PKL_NAME = "gdv.pkl"
OUT_DIR = RESULTS_BASE / "aggregates"
OUT_DIR.mkdir(parents=True, exist_ok=True)

def load_labels_from_pkl(pkl_path: Path):
    """Return label list for this persona.
    Prefer meta['labels']; else take first layer's 'group'."""
    with open(pkl_path, "rb") as f:
        obj = pickle.load(f)
    # try meta
    labels = None
    meta = obj.get("meta", {})
    if isinstance(meta, dict):
        labels = meta.get("labels")
    # fallback to first layer's 'group'
    if not labels:
        layer_data = obj.get("layer_data", {})
        sorted_layers = obj.get("sorted_layers") or list(layer_data.keys())
        if sorted_layers:
            first_key = sorted_layers[0]
            labels = layer_data[first_key].get("group")
    if labels is None:
        raise ValueError(f"Could not find labels in {pkl_path}")
    return list(labels)

# 1) Load all persona label lists
persona_labels = {}
missing = []
for i in range(10):
    pkl_path = RESULTS_BASE / PERSONA_DIR_PATTERN.format(i=i) / PKL_NAME
    if not pkl_path.exists():
        missing.append(str(pkl_path))
        continue
    try:
        labels = load_labels_from_pkl(pkl_path)
        persona_labels[f"persona_{i}"] = labels
    except Exception as e:
        print(f"[WARN] {pkl_path}: {e}")

if missing:
    print(f"[INFO] Missing PKLs ({len(missing)}):")
    for m in missing:
        print("  -", m)

if not persona_labels:
    raise RuntimeError("No persona PKLs were loaded. Check paths.")

# 2) Align lengths (truncate to common min length to keep indices consistent)
lengths = {k: len(v) for k, v in persona_labels.items()}
min_len = min(lengths.values())
max_len = max(lengths.values())
if min_len != max_len:
    print(f"[WARN] Label list lengths differ across personas. Truncating to min length = {min_len}")
    for k in persona_labels:
        persona_labels[k] = persona_labels[k][:min_len]

# 3) Save persona → labels JSON
persona_json_path = OUT_DIR / "persona_labels.json"
with open(persona_json_path, "w", encoding="utf-8") as f:
    json.dump(persona_labels, f, ensure_ascii=False, indent=2)
print(f"[OK] Saved persona labels → {persona_json_path}")

# 4) Per-image label counts across personas
personas = sorted(persona_labels.keys(), key=lambda x: int(x.split("_")[1]))
n_images = min_len
image_label_counts = []  # list[dict] of label->count per image index
majority_vote = []       # list[str] majority label per image (ties broken by label name)

all_labels_seen = set()
for idx in range(n_images):
    lab_list = [persona_labels[p][idx] for p in personas]
    c = Counter(lab_list)
    image_label_counts.append(dict(c))
    all_labels_seen.update(c.keys())
    # majority vote (deterministic tie-break: highest count, then alphabetically)
    top_count = max(c.values())
    winners = sorted([lab for lab, cnt in c.items() if cnt == top_count])
    majority_vote.append(winners[0])

# 5) Save image label counts (JSON)
counts_json_path = OUT_DIR / "image_label_counts.json"
with open(counts_json_path, "w", encoding="utf-8") as f:
    json.dump(image_label_counts, f, ensure_ascii=False, indent=2)
print(f"[OK] Saved per-image label counts → {counts_json_path}")

# 6) Also save a CSV (long-form) with majority vote
import csv
counts_csv_path = OUT_DIR / "image_label_counts.csv"
with open(counts_csv_path, "w", newline="", encoding="utf-8") as f:
    w = csv.writer(f)
    w.writerow(["image_index", "label", "count", "majority_label"])
    for idx, counts in enumerate(image_label_counts):
        maj = majority_vote[idx]
        # ensure rows for all labels seen so columns aren't implied
        for lab in sorted(all_labels_seen):
            w.writerow([idx, lab, counts.get(lab, 0), maj])
print(f"[OK] Saved per-image label counts CSV → {counts_csv_path}")

# 7) Quick sanity print
print("\n=== Summary ===")
print(f"Personas loaded: {len(personas)} → {personas}")
print(f"Images considered: {n_images}")
print(f"Unique labels observed: {sorted(all_labels_seen)}")
print("First 5 images' majority votes:", majority_vote[:5])
```
